### Un algoritmo genético para la asignación de recursos académicos a estudiantes
#### Generación de catálogos de recursos y perfiles de estudiante

Este programa realiza los siguientes pasos:

- Genera aleatoriamente un catálogo de recursos académicos de tamaño $n\gt 0$, donde cada recurso tiene un tiempo $\tau_i \in \mathbb{N}_{+}$, con $t_{min}\leq\tau_i\leq t_{max}$, $i=1,\ldots, n$. Los valores $n$, $t_{min}$ y $t_{max}$ son definidos por el usuario.

- Guarda el catálogo de recursos en un fichero .csv.

- Genera aleatoriamente un perfil de estudiante con identificador y un diagnóstico definidos por el usuario.

- Guarda el perfil de estudiante en un fichero .csv.

**Detalles en:**

Ana F. Hernández, Andrej Franulic, Fernando Jiménez. A genetic algorithm for student academic resource allocation. Technical Report Erasmus+ KA220-SCH – Cooperation Partnerships in School Education, July 2026.

##### Autor:
Fernando Jiménez Barrionuevo

```python
import numpy as np
import pandas as pd
import time
import os
import matplotlib.pyplot as plt
from itertools import product

# Información para la generación de catálogos de recursos
AREAS = ["Numbers", "Algebra", "Geometry", "Statistics", "Functions"]
COURSES = [1, 2, 3, 4]
LEVELS = [1, 2, 3]
TYPES = ["Theory", "Practice"]

# Información para la generación de perfiles de estudiantes
DIAGNOSTICS = {
    "AACC": 1.0,
    "Ordinario": 0.8,
    "Dislexia": 0.7,
    "TDAH": 0.6,
    "Discalculia": 0.5
}

# Semilla para el generador de números aleatorios (para reproducibilidad)
seed = 1

# Parámetros de usuario para la generación del catálogo y del perfil de estudiante
n = 1000 # tamaño del catálogo
tmin = 60 # tiempo mínimo de los recursos en minutos
tmax = 240 # tiempo máximo de los recursos en minutos
ID = 1 # identificador del estudiante 
diagnostic = "Dislexia" # diagnóstico del estudiante

# Función para la generación de catálogos
def generate_catalog(n, tmin, tmax, seed):
    """
    Genera un catálogo aleatorio de n recursos con un tiempo mínimo tmin y un timepo máximo tmax (en minutos) para cada recurso.
    """
    np.random.seed(seed)
    catalog = []
    while len(catalog) < n:
        resource = {
            'id': len(catalog),
            'course': np.random.choice(COURSES),
            'area': np.random.choice(AREAS),
            'level': np.random.choice(LEVELS),
            'type': np.random.choice(TYPES),
            'time': np.random.randint(tmin, tmax)
        }
        catalog.append(resource)
    return catalog

# Función para la generación de perfiles de estudiante modificada
def generate_student_profile(student_id, diagnosis, seed):
    """
    Genera un perfil de estudiante con un identificador, diagnóstico y campos individuales de GE por área.
    """
    np.random.seed(seed)
    course = int(np.random.choice(COURSES))
    
    # Crear el diccionario base con los datos generales
    profile = {
        'student_id': student_id,
        'course': course,
        'diagnostic': diagnosis,
        's': DIAGNOSTICS[diagnosis]
    }
    
    # Rellenar dinámicamente solo los 5 campos de GE (del 1 al 15 inclusive)
    for area in AREAS:
        ge = int(np.random.randint(1, 16))
        profile[f'GE_{area}'] = ge
        
    return profile

# Generar catálogo de recursos
print("Generando catálogo...")
catalogue = generate_catalog(n,tmin,tmax,seed)

# Convertir catalogo a DataFrame y guardar en CSV
df_catalogue = pd.DataFrame(catalogue)
df_catalogue.to_csv('catalogue.csv', index=False, encoding='utf-8')
print("Catálogo guardado con éxito en catalogue.csv")

# Generar perfil de estudiante
print("Generando perfil de estudiante...")
student = generate_student_profile(ID,diagnostic,seed)

# Convertir perfil de estudiante a DataFrame y guardar en CSV
df_student = pd.DataFrame([student])
df_student.to_csv('student_profile.csv', index=False, encoding='utf-8')
print("Perfil de estudiante con éxito en student_profile.csv")

print("\nEjecución completada correctamente.")
```
